```python
import pandas as pd
from pymongo import MongoClient
import os
```

```python
client = MongoClient('mongodb://localhost:27017/')
db = client['StockMarketDatabase']
collection = db['StockMarketCollection']
```

Loading stocks data into collection

```python
folder_path = 'C:/Users/almas/Downloads/archive17/Data/Stocks'
```

```python
# Iterate through files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)

        # Read CSV file into pandas DataFrame
        df = pd.read_csv(file_path)

        # Check if the DataFrame is empty
        if df.empty:
            print(f"Skipping empty file: {filename}")
            continue

        # Extract the ticker symbol from the filename
        ticker_symbol = filename.split('.')[0]

        # Add the 'Ticker' column to the DataFrame
        df['Ticker'] = ticker_symbol

        # Convert pandas DataFrame to list of dictionaries
        data = df.to_dict('records')

        # Insert data into MongoDB collection
        collection.insert_many(data)
```
